Async io is the python module that helps write consurrent programs on a single thread and using an event loop

Synchronous code is where everything runs one after another, whereas in concurrent code the task that is taking some time is paused and another task is started. 

Remember concurrency doesnt really mean faster in terms of processing speed, its just that during the time the program was waiting earlier, its now just executing another part of the program

Asyncio excels in io bound tasks. Asyncio uses what is called cooperative multitasking, where a task voluntarily gives up control

For CPU bound tasks which uses heavy computation we would use "processes" instead

### Event Loop:
Event loop is the main engine that controls the program execution. Think of it as a scheduler which keeps track of all the tasks and as soon as a task is suspended it starts some other task

event loop is started with the code

asyncio.run(function_name())

## await 

awaitables are objects that implement special __await__() method under the hood. An object has to be awaitable to use await keyword on it

We cant use await keyword with synchronous functions because they dont have the mechanism to interact with event loop, i.e. they cant yielc control to the event loop, and we cant use await kewyword for them. Therefore we can not use time.sleep() in async because its a synchronous function and we have to use asyncio.sleep()

We can use await keyword only inside an async function i.e. the functions defined with async keyword before them 

Await keyword basically asks a function to pause and yield control to event loop which can then execute another task. The suspended function will remain suspended until the awaitable completes.


There are 3 types of awaitables in asyncio: coroutines, tasks, futures

1. Coroutines: A function defined with "async def" creates a coroutine object. Calling it does not run the body, it has to be awaited or scheduled

2. Tasks: are wrappers around the coroutines which schedules tasks to run concurrently 

3. Futures: are low level objects representing results
